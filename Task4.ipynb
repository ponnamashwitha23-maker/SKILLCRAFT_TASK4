```python
import pandas as pd
import zipfile
import os

# Path to the zip file
zip_file_path = "/content/archive (3).zip"

# Check if the zip file exists and unzip it
if os.path.exists(zip_file_path):
    print(f"Unzipping '{zip_file_path}'...")
    with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
        zip_ref.extractall(".") # Extract all contents to the current directory
    print("Unzipping complete.")
else:
    print(f"Warning: '{zip_file_path}' not found. Please ensure the zip file is uploaded correctly.")

# Now, try to read the CSV file with the correct filename
df = pd.read_csv("Marketing campaign dataset.csv")

df.head()
```
